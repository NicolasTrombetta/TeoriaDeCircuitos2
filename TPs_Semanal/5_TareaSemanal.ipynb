$$

Ahora aplicamos la particion de Krön, para la supresion del nodo V4.

$$
[Y']=[Y_{11}]−[Y_{12}][Y_{22}]^{-1}[Y_{21}]
$$

donde:
$$
Y_{11}= \begin{bmatrix}
\frac{1}{SL}  &      0        & 0   \\
       0      & \frac{1}{SL}  & 0   \\
       0      &      0        & SC  \\
\end{bmatrix},
\qquad
Y_{12}= \begin{bmatrix}
 -\frac{1}{SL} \\
 -\frac{1}{SL} \\
     -SC        \\
\end{bmatrix},
\qquad
Y_{21}= \begin{bmatrix}
-\frac{1}{SL} & -\frac{1}{SL} & -SC  
\end{bmatrix},
\qquad
Y_{22}=\begin{bmatrix}
\frac{S^2LC+2}{SL}
\end{bmatrix}
$$

$$

#### MAI de mosfet



Planteando Nudos:
$$V_s[S(C_1+C_3)]-V_g(SC_1)-V_d(SC_3)=gm(V_g-V_s) => V_s[S(C_1+C_3)+gm]-V_g(SC_1+gm)-V_d(SC_3)=0$$
$$V_g[S(C_1+C_2)]-V_d(SC_2)-V_s(SC_1)=0$$
$$V_d[S(C_3+C_2)]-V_s(SC_3)-V_g(SC_2)=-gm(V_g-V_s) => V_d[S(C_3+C_2)]-V_s(SC_3+gm)-V_g(SC_2+gm)=0 $$

Tomando $V_s$ como nodo 1, $V_g$ como nodo 2 y $V_d$ como nodo 3 la MAI queda:
$$
Y = \begin{bmatrix}
S(C_1+C_3)+gm  &  -(SC_1+gm)   & -SC_3   \\
-SC_1          &  S(C_1+C_2)   & -SC_2   \\
-(SC_3 + gm)   &  -(SC_2-gm)   & S(C_3+C_2)   \\
\end{bmatrix}
$$

Donde $C_1+C_2=C_{iss}$, $C_3+C_2=C_{oss}$ y $C_2=C_{rss}$

b ) Graficar y observar detenidamente la respuesta de módulo y fase para cada transferencia. Analizar si entre 2 de las transferencias halladas se cumple que son complementarias en módulo : $|T_a + T_b| = 1$

Recordando que: 
$$
{A_v}^{ij}_{mn}=\frac{V_{ij}}{V_{mn}}=(-1)^{i+j+m+n}\frac{\underline{Y}^{mn}_{ij}}{\underline{Y}^{mn}_{mn}}
$$

Entonces:
$$
{A_v}^{23}_{13} = \frac{V_{23}}{V_{13}}=-\frac{\underline{Y}^{13}_{23}}{\underline{Y}^{13}_{13}},
\qquad
{A_v}^{21}_{31} = \frac{V_{21}}{V_{31}}=-\frac{\underline{Y}^{31}_{21}}{\underline{Y}^{31}_{31}},
\qquad
{A_v}^{12}_{32} = \frac{V_{12}}{V_{32}}=\frac{\underline{Y}^{32}_{12}}{\underline{Y}^{32}_{32}}
$$

donde: $$
\underline{Y}^{13}_{23}=-\frac{1}{SL(S^2LC+2)}, 
\qquad
\underline{Y}^{13}_{13}= \underline{Y}^{31}_{31}= \underline{Y}^{32}_{32}=\frac{S^2LC+1}{SL(S^2LC+2)},
\qquad
\underline{Y}^{31}_{21}=\underline{Y}^{32}_{12}=-\frac{SC}{S^2LC+2},
$$

Finalmente: 
$$
{A_v}^{23}_{13} = \frac{V_{23}}{V_{13}}=(signo)\frac{-1}{S^2LC+1}
$$
$$
{A_v}^{21}_{31} = \frac{V_{21}}{V_{31}}=(signo)\frac{S^2LC}{S^2LC+1}
$$
$$
{A_v}^{12}_{32} = \frac{V_{12}}{V_{32}}=(signo)\frac{S^2LC}{S^2LC+1}
$$

```python
from scipy import signal
import numpy as np
from matplotlib import pyplot as plt

L=1
C=1

T_23_13=signal.TransferFunction([-1],[L*C,0,1])  
T_21_31=signal.TransferFunction([L*C,0,0],[L*C,0,1])
T_12_32=signal.TransferFunction([-L*C,0,0],[L*C,0,1])

# Frecuencias
w = np.logspace(-2, 2, 1000)

# Magnitud de cada función
w1, mag1, phase1 = signal.bode(T_23_13, w=w)
w2, mag2, phase2 = signal.bode(T_21_31, w=w)
w3, mag3, phase3 = signal.bode(T_12_32, w=w)

# Gráfico
plt.figure(figsize=(8, 5))

plt.semilogx(w1, mag1, label='T_23_13')
plt.semilogx(w2, mag2, label='T_21_31')
plt.semilogx(w3, mag3, label='T_12_32')

plt.xlabel('Frecuencia angular [rad/s]')
plt.ylabel('Magnitud [dB]')
plt.title('Magnitud de las funciones de transferencia')
plt.grid(True, which='both')
plt.legend()

plt.show()



```
