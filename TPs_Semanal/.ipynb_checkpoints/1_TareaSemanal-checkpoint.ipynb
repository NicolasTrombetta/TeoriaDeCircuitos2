Para hallar la transferencia $H(s)=\frac{V_2}{V_1}$ del circuito planteamos nodos,donde $V^{+}$ y $V^{-}$ denotan la tension en la pata positiva y negativa del opam respectivamente.:
\begin{equation}\tag{1}
V^{-} (G_1+G_2)-V_1 G_1-V_2 G_2 = 0 
\end{equation}

\begin{equation}\tag{2}
V^{+} (S C_1+G_3)-V_1 S C_1 = 0
\end{equation}

Para simplificar los calculos asumimos que el circuito esta correctamente realimentado y por lo tanto hay un cortocircuito virtual en la terminales de entrada del OPAM,es decir, $V^{-} =V^{+}=V_A$ 

Despejando $V_A$ de $(1)$ y $(2)$ e igualando obtenemos:
\begin{equation}
\frac{V_1 G_1+V_2 G_2}{(G_1+G_2)} = \frac{V_1 S C_1}{(S C_1+G_3)} 
\end{equation}

Finalmente obtenemos:
\begin{equation}
T(s)=\frac{V_2}{V_1}=\frac{S-\frac{R_2}{R_1 R_3 C_1}}{S+\frac{1}{R_3 C_1}}
\end{equation}

Si $R_2=R_1$ entonces:
\begin{equation}
T(s)=\frac{V_2}{V_1}=\frac{S-\frac{1}{R_3 C_1}}{S+\frac{1}{R_3 C_1}}
\end{equation}

#### Diagrama de Polos y Ceros


#### Diagrama de modulo y fase

Para el diagrama de modulo y fase reemplazamos $S=jw$:
\begin{equation}
T(jw)=\frac{jw-\frac{1}{R_3 C_1}}{jw+\frac{1}{R_3 C_1}}
\end{equation}

De donde se deduce que:
\begin{equation}
|T(jw)|=1
\end{equation}
\begin{equation}
\theta(jw)=arctg(\frac{w}{\frac{-1}{R_3 C_1}})-arctg(\frac{w}{\frac{1}{R_3 C_1}})
\end{equation}

Reescribimos la fase para graficarla mas facil:
Para $\omega > 0$ $ arctg(\frac{\omega}{-1/R_3C_1})$  esta en el segundo cuadrante por lo tanto 
$$
 arctg(\frac{\omega}{-1/R_3C_1})= \pi+arctg(-\omega C_1R_3)=\pi-arctg(\omega C_1R_3)
$$
entonces la fase para $\omega > 0$ queda:
$$
\theta(jw)=\pi-2arctg(\omega C_1R_3)
$$




Ahora buscamos la respuesta de retardo de grupo:
$$
\tau_g(\omega) = -\frac{d\phi(\omega)}{d\omega}
$$
$$
\tau_g(\omega)=\frac{2}{R_3C_1}\frac{1}{1+(R_3C_1\omega)^2}
$$

Se trata de un filtro pasatodo o retardador de fase.Se utiliza para modificar la fase de una señal sin alterar su amplitud,por ejemplo para introducir retardos controlados en sistemas de audio y comunicaciones sin cambiar el balance tonal.

### Normalizacion
Para la normalizacion de los compenentes propongo tomar una $\Omega_z=R_3$ y $\Omega_\omega=\frac{1}{R_3 C_1}$ entonces los valores de los componentes normalizados quedan:
\begin{equation}
R_3^{'}=1
\end{equation}
\begin{equation}
C_1^{'}=1
\end{equation}
Y la trasferencia:
$$
T(S)=\frac{S-1}{S+1}
$$






### Simulacion numérica de la función transferencia normalizada en Python.

```python
from scipy import signal as sig
import matplotlib as mpl
from matplotlib import pyplot as plt
from pytc2.sistemas_lineales import bodePlot, pzmap, GroupDelay, analyze_sys

num=[1,-1]
den=[1,1]

T=sig.TransferFunction(num,den)

pzmap(T)
_, axes_hdl = bodePlot(T)

plt.figure()
plt.sca(axes_hdl[0])
plt.ylim([-1, 1])
plt.gca
plt.show()
```

## Simulacion con sympy sin asumir realimentacion negativa

```python
import sympy as sp
from pytc2.general import print_latex

# símbolos
V1, V2, V_mas, V_menos= sp.symbols('V1 V2 V_mas V_menos')
R1, R2, R3, C1 = sp.symbols('R1 R2 R3 C1')
S, As=sp.symbols('S As')

eqs=[
    V_mas*(1/R1+1/R2)-V1*(1/R1)-V2*(1/R2),
    V_menos*(S*C1+1/R3)-V1*S*C1,
    As*(V_mas-V_menos)-V2
]

sol=sp.solve(eqs,[V1,V2,V_mas,V_menos])

T0=sol[V2]/sol[V1]
display(Math("T = " + sp.latex(T0)))


```

Para simplificar la expresion hacemos $As\to \infty$

```python
T=sp.limit(T0,As,sp.oo)
from IPython.display import display, Math
import sympy as sp

display(Math("T = " + sp.latex(T)))
```

Si ordenamos la expresion nos devuelve python,podemos observar que coincide con la calculada anteriormente.Por lo tanto, esta bien asumida la realimentacion negativa.
