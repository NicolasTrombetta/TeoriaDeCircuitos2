$$
con $R_g=1$

$$
Z_{21}=\frac{s(s^2+\frac{1}{16})}{2s^2+1}
$$
$$
Z_{11}=\frac{s(s^2+2)}{2s^2+1}
$$

$$

donde $$k_1=\lim_{s^2 \to -\frac{1}{16}}\frac{s^2+\frac{1}{16}}{s}Z_1(s)$$

Entonces:
$$
Z_3=\frac{31}{48}s
$$
