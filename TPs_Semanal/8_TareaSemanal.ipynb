1) Dada una determinada función impedancia $Z(s)$ se sabe que:
- La pulsación libre de amortiguamiento y el coeficiente de amortiguamiento asociados a sus ceros están dados por $1\frac{rad}{s}$ y $0.5\frac{rad}{s}$ .

- Posee polos en $S=-1$ y $S=5e^{+j116.6°}$respectivamente. 

Determine el valor de los componentes que integran el siguiente dipolo, sabiendo que satisface la impedancia propuesta.



Empezamos por escribir la expresion de Z(s):
$$
Z(s)=k\frac{s^2+s+1}{(s+1)(s^2+4.48s+25)}
$$

Removemos el capcitor $C_1$, es decir, se remueve un Admitancia en infinito.
$$
Y_2(s)=Y(s)-Y_1(s)=\frac{1}{k}\frac{s^3+5.48s^2+29.48s+25}{s^2+s+1}-k_\infty s
$$
donde:
$$
k_\infty=\lim_{s \to \infty}\frac{Y(s)}{s}=\frac{1}{k}
$$
Entonces:
$$
Y_2(s)=\frac{1}{k}\frac{1}{k}\frac{s^3+5.48s^2+29.48s+25}{s^2+s+1}-\frac{1}{k}\frac{s^3+s^2+s}{s^2+s+1}
$$
$$
Y_2(s)=\frac{1}{k}\frac{4.48s^2+28.48s+25}{s^2+s+1}
$$

Removemos ahora $R_1$:
$$
Y_4(s)=Y_2(s)-G_1
$$
donde $$G_1=\lim_{s \to \infty}Y_2(s)=\frac{4.48}{k}$$
entonces:
$$
Y_4(s)=\frac{1}{k}\frac{4.48s^2+28.48s+25}{s^2+s+1}-\frac{4.48}{k}
$$
$$
Y_4(s)=\frac{1}{k}\frac{24s+29.48}{s^2+s+1}
$$

Removemos $L_1$:
$$
Z_6(s)=Z_4(s)-sL1
$$

Donde: $L_1=\lim_{s \to \infty}\frac{Z_4(s)}{s}=\frac{k}{24}$

Entonces:
$$
Z_6(s)=k\frac{0.145s+1}{24s+20.52}
$$

Removemos R2:
$$
R_2=\lim_{s \to \infty}Z_6(s)=0.006k
$$
Entonces la impedancia $Z_8$ queda:
$$
Z_8(s)=Z_6(s)-R_2=k\frac{0.876}{24s+20.52}
$$

Finalmente lo que queda es $C_2=0.0365k$ y $R_3=\frac{32.043}{k}$
